# **1. Handling Missing (Null) Values**

```python
import pandas as pd

data = {'Name': ['John', 'Anna', None],
        'Age': [25, None, 30]}
df = pd.DataFrame(data)

print("Original DataFrame:\n", df)
print("\nNull values in DataFrame:\n", df.isnull())
```

# **2. Filling Missing Values**

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Name'] = df['Name'].fillna("Unknown")


print("After filling missing values:\n", df)
```

# **3. Outliers Detection and Removal (IQR)**

```python
df = pd.DataFrame({'Salary': [25000, 26000, 27000, 1000000]})
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

filtered_df = df[(df['Salary'] >= Q1 - 1.5 * IQR) & (df['Salary'] <= Q3 + 1.5 * IQR)]

print("Original Salaries:\n", df)
print("\nFiltered Salaries (Outliers Removed):\n", filtered_df)
```

# **4. Feature Engineering – Creating New Columns**

```python
df = pd.DataFrame({'Weight': [60, 70], 'Height': [1.6, 1.75]})
df['BMI'] = df['Weight'] / (df['Height'] ** 2)
print("BMI Calculation:\n", df)
```

# **5. Removing Unwanted Columns**

```python
df = pd.DataFrame({'ID': [1, 2], 'Name': ['A', 'B'], 'Age': [25, 30]})
df.drop('ID', axis=1, inplace=True)
print("After removing 'ID' column:\n", df)
```

# **6. Creating New Columns**

```python
df = pd.DataFrame({'BirthYear': [2000, 1995]})
df['Age'] = 2025 - df['BirthYear']
print("Age from BirthYear:\n", df)
```

# **7. Converting Column Types**

```python
df = pd.DataFrame({'Salary': ['1000', '2000']})
df['Salary'] = df['Salary'].astype(int)
print("Converted Salary Type:\n", df.dtypes)
```

# **8. Converting to DateTime**

```python
df = pd.DataFrame({'JoinDate': ['2024-01-01', '2023-05-12']})
df['JoinDate'] = pd.to_datetime(df['JoinDate'])
print("Converted to datetime:\n", df)
```

# **9. Renaming Columns**

```python
df = pd.DataFrame({'emp_name': ['Alice'], 'emp_age': [28]})
df.rename(columns={'emp_name': 'Name', 'emp_age': 'Age'}, inplace=True)
print("Renamed Columns:\n", df)
```
