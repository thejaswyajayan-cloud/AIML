```python
# Total profit for the year
total_profit = company_sales_df["total_profit"].sum()
print("Total Yearly Profit:", total_profit)
```

```python
# Maximum selling product by total units
product_totals = company_sales_df[["facecream", "facewash", "toothpaste", "bathingsoap", "shampoo", "moisturizer"]].sum()
print("Product-wise Total Sales:\n",product_totals)
```

# **LINE CHART**

```python
plt.figure(figsize=(10, 6))# Creates a new figure with a specified width and height (10 inches by 6 inches) for better visibility

plt.plot(company_sales_df["month_number"], company_sales_df["facecream"], label="Face Cream", marker='o')
plt.plot(company_sales_df["month_number"], company_sales_df["facewash"], label="Face Wash", marker='o')
plt.plot(company_sales_df["month_number"], company_sales_df["toothpaste"], label="Toothpaste", marker='o')
plt.plot(company_sales_df["month_number"], company_sales_df["bathingsoap"], label="Bathing Soap", marker='o')
plt.plot(company_sales_df["month_number"], company_sales_df["shampoo"], label="Shampoo", marker='o')
plt.plot(company_sales_df["month_number"], company_sales_df["moisturizer"], label="Moisturizer", marker='o')

plt.title("Monthly Sales Data")
plt.xlabel("Month")# Labels the x-axis as "Month"
plt.ylabel("Units Sold")# Labels the y-axis as "Units Sold"
plt.legend()# Displays the legend using the labels given in the plot() functions
plt.grid(True)# Adds a grid to the background of the plot to improve readability
plt.tight_layout()# Automatically adjusts plot elements to prevent overlap or clipping
plt.show()# Displays the final plot in the output cell or window
```

# **BAR PLOT**

```python
product_totals.plot(kind='bar', figsize=(8,5), color='skyblue')     # ➤ Creates a vertical bar chart with light blue bars and sets figure size
plt.title("Total Sales per Product - Yearly")                        # ➤ Sets the chart title
plt.ylabel("Units Sold")                                             # ➤ Labels the Y-axis
plt.xlabel("Product")                                                # ➤ Labels the X-axis
plt.xticks(rotation=45)                                              # ➤ Rotates X-axis labels for better readability
plt.grid(axis='y')                                                   # ➤ Adds horizontal grid lines for Y-axis to improve readability
plt.tight_layout()                                                   # ➤ Adjusts layout to prevent overlap
plt.show()                                                           # ➤ Displays the final plot

```

# **PIE CHART**

```python
# Plotting the pie chart
plt.figure(figsize=(6, 6))                                            # ➤ Sets the figure size to 8x8 inches for a square plot

plt.pie(product_totals,labels=product_totals.index, autopct='%1.1f%%',startangle=140)                                                   # ➤ Data values (total sales per product)labels=product_totals.index,                                      # ➤ Labels (product names)
                                                      # ➤ Starts the pie chart from 140° for better layout

plt.title("Total Sales Distribution (Product-wise)")                  # ➤ Sets the title of the pie chart

plt.axis("equal")                                                     # ➤ Ensures the pie chart is a perfect circle

plt.tight_layout()                                                    # ➤ Adjusts layout to avoid overlapping elements

plt.show()                                                            # ➤ Displays the final pie chart
```

```python
# Histogram of total profit
plt.figure(figsize=(8,6))
plt.hist(company_sales_df["total_profit"], bins=6, color="skyblue", edgecolor="black")

plt.xlabel("Total Profit", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.title("Histogram of Total Profit", fontsize=14)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
