```python
#question 1
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_employee(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def display_manager(self):
        self.display_employee()   # Accessing parent class method
        print("Department:", self.department)

m = Manager("Rahul", 50000, "CSE")

print("Name:", m.name)
print("Salary:", m.salary)

m.display_employee()
m.display_manager()
```

```python
#question 3
import pandas as pd
import numpy as np

# Generate random marks between 60 and 95
np.random.seed(10)

marks = np.random.randint(60, 96, 5)

data = {
    "Roll Number": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Rahul", "Meera", "Arjun", "Neha"],
    "Department": ["CSE", "ECE", "IT", "CSE", "EEE"],
    "Marks": marks
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
```

```python
#Question 3(A)
print("Students scoring more than 75:")
print(df[df["Marks"] > 75])
```

```python
#Question 3(B)
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(calculate_grade)

print(df)
```

```python
#Question 3(C)
df_sorted = df.sort_values(by="Marks", ascending=False)

print("DataFrame sorted by Marks:")
print(df_sorted)
```
